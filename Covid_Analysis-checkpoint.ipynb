```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/SR1608/Datasets/main/covid-data.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.shape
df.dtypes
df.info()
df.describe()
```

```python
df['location'].nunique()
df['continent'].value_counts()
df['total_cases'].max()
df['total_cases'].mean()
df['total_deaths'].quantile([0.25,0.50,0.75])
df.groupby("continent")['human_development_index'].max()
df.groupby("continent")['gdp_per_capita'].min()
```

```python
df = df[['continent','location','date','total_cases','total_deaths','gdp_per_capita','human_development_index']]
df.head()
```

```python
df.drop_duplicates(inplace=True)
df.isnull().sum()
df = df.dropna(subset=['continent'])
df = df.fillna(0)
```

```python
df['date'] = pd.to_datetime(df['date'], format="%Y-%m-%d", errors='coerce')
df['month'] = df['date'].dt.month
df.head()
```

```python
df_groupby = df.groupby("continent").max().reset_index()
df_groupby
```

```python
df_groupby['total_deaths_to_total_cases'] = df_groupby['total_deaths'] / df_groupby['total_cases']
df_groupby
```

```python
sns.histplot(df['gdp_per_capita'])
plt.show()
sns.scatterplot(x=df['total_cases'], y=df['gdp_per_capita'])
plt.show()
sns.pairplot(df_groupby)
plt.show()
sns.catplot(data=df_groupby, x='continent', y='total_cases', kind='bar')
plt.show()
```

```python
df_groupby.to_csv("df_groupby.csv", index=False)
```
